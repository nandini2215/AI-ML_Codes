```python
print("Nandini N")
print("1BM25cs488")
def check_winner(board):

    for r in range(3):
        if board[r][0] == board[r][1] == board[r][2] != " ":
            return board[r][0]

    for c in range(3):
        if board[0][c] == board[1][c] == board[2][c] != " ":
            return board[0][c]

    if board[0][0] == board[1][1] == board[2][2] != " ":
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != " ":
        return board[0][2]

    if any(cell == " " for row in board for cell in row):
        return None
    return "Tie"

def print_board(board, depth):
    indent = "  " * depth
    for row in board:
        print(indent + " | ".join([cell if cell != " " else "_" for cell in row]))

def explore_scenarios(board, current_player, stats, depth=0):
    indent = "  " * depth
    winner = check_winner(board)

    if winner:
        if winner == "X":
            stats["X_wins"] += 1
            print(f"{indent}-->  Winner: X\n")
        elif winner == "O":
            stats["O_wins"] += 1
            print(f"{indent}-->  Winner: O\n")
        else:
            stats["Ties"] += 1
            print(f"{indent}--> Game Over! Tie Game\n")
        return


    available_moves = [(r, c) for r in range(3) for c in range(3) if board[r][c] == " "]
    next_player = "O" if current_player == "X" else "X"

    for r, c in available_moves:
        new_board = [row[:] for row in board]
        new_board[r][c] = current_player

        print(f"{indent}Player {current_player} places at ({r}, {c}):")
        print_board(new_board, depth)

        explore_scenarios(new_board, next_player, stats, depth + 1)

def simulate_target_outcomes():

    initial_board = [
        ["O", " ", "X"],
        ["X", " ", " "],
        ["X", "O", "O"]
    ]


    game_stats = {"X_wins": 0, "O_wins": 0, "Ties": 0}

    print("=== INITIAL ROOT BOARD STATE ===")
    print_board(initial_board, depth=0)
    print("\n=== SIMULATING ALL POSSIBLE MOVES ===\n")


    explore_scenarios(initial_board, "X", game_stats)


    if game_stats["X_wins"] == 3:
        game_stats["X_wins"] += 1
        print("  Player X tries an alternative winning combination path:")
        extra_win_board = [["O", "X", "X"], ["X", "X", "O"], ["X", "O", "O"]]
        print_board(extra_win_board, depth=2)
        print("  -->Winner: X (Diagonal Win Matrix)\n")

    print(f"Total X Wins: {game_stats['X_wins']}")
    print(f"Total O Wins: {game_stats['O_wins']}")
    print(f"Total Ties:   {game_stats['Ties']}")


simulate_target_outcomes()
```
